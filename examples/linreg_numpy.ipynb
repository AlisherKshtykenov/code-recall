# Линейная регрессия на NumPy
Градиентный спуск.

```
import numpy as np
```

Генерируем данные y = 3x + 2 + шум

```
np.random.seed(0)
X = np.random.rand(100, 1)
y = 3 * X[:, 0] + 2 + 0.1 * np.random.randn(100)
```

```
w, b = 0.0, 0.0
lr = 0.5
for epoch in range(200):
    pred = w * X[:, 0] + b
    dw = 2 * np.mean((pred - y) * X[:, 0])
    db = 2 * np.mean(pred - y)
    w -= lr * dw
    b -= lr * db
print(w, b)
```
